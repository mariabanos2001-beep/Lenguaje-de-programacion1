# Parte 2: Diccionario portafolio
Se construye un diccionario donde cada cliente es una clave y tiene como valor
un subdiccionario con su `monto_invertido` y `rendimiento`.  
Después, se imprimen los clientes con rendimiento mayor al 5%.


```python
# Parte 2: Diccionario portafolio
portafolio = {
    "Juan Perez": {"monto_invertido": 10000, "rendimiento": 5.5},
    "Maria Rodriguez": {"monto_invertido": 15000, "rendimiento": 6.0},
    "Carlos Lopez": {"monto_invertido": 20000, "rendimiento": 4.0},
    "Ana Gomez": {"monto_invertido": 18000, "rendimiento": 7.5},
}

# Evaluación: imprimir clientes con rendimiento > 5%
for cliente, datos in portafolio.items():
    if datos["rendimiento"] > 5:
        print(f"{cliente} tiene un rendimiento de {datos['rendimiento']}%")
```

# Parte 3: Cálculo del rendimiento total del portafolio
Se calcula el rendimiento de cada cliente en **dinero**, se suman y se obtiene
el rendimiento total.  

Luego, se identifica el cliente que genera el mayor rendimiento absoluto
(en dólares).


```python
# Parte 3: Cálculo del rendimiento total del portafolio
rendimiento_total = 0
for cliente, datos in portafolio.items():
    rendimiento_cliente = (datos["monto_invertido"] * datos["rendimiento"]) / 100
    rendimiento_total += rendimiento_cliente

print(f"El rendimiento total del portafolio es: ${rendimiento_total:.2f}")

# Identificar cliente con mayor rendimiento absoluto (en dólares)
max_rend = -1
cliente_max = None
for cliente, datos in portafolio.items():
    rendimiento_cliente = (datos["monto_invertido"] * datos["rendimiento"]) / 100
    if rendimiento_cliente > max_rend:
        max_rend = rendimiento_cliente
        cliente_max = cliente

print(f"Cliente con mayor rendimiento absoluto: {cliente_max} -> ${max_rend:.2f}")
```

# Parte 4: Modificaciones del portafolio
Aquí se realizan tres cambios:  
1. Se **actualiza** el monto invertido de Carlos Lopez a $25,000.  
2. Se **añade** a una nueva clienta: Laura Fernandez.  
3. Se **elimina** a Ana Gomez.  

Finalmente, se imprime el portafolio actualizado.


```python
# Parte 4: Modificaciones del portafolio
# 1) Actualizar Carlos Lopez
portafolio["Carlos Lopez"]["monto_invertido"] = 25000

# 2) Añadir Laura Fernandez
portafolio["Laura Fernandez"] = {"monto_invertido": 22000, "rendimiento": 6.8}

# 3) Remover Ana Gomez
if "Ana Gomez" in portafolio:
    del portafolio["Ana Gomez"]

# Portafolio actualizado
print("\nPortafolio actualizado:")
for cliente, datos in portafolio.items():
    print(f"- {cliente}: monto={datos['monto_invertido']}, rendimiento%={datos['rendimiento']}")
```
